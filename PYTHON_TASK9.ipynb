```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("profit_sales_analysis_dataset.csv")
```

```python
df
```

```python
numeric_cols=["Unit_Price","Quantity","Discount_Percent","Sales","Profit","Profit_Margin_Percent","Cost"]
numeric=df[numeric_cols].corr()
numeric
```

```python
#heatplot
sns.heatmap(numeric,
            annot=True,
            cmap="coolwarm",
            cbar=False,
            square=True,
            linewidth=1)
plt.title("profit_margin")
plt.show()
```

```python
#bar plot
sns.barplot(x="Category",y="Sales",data=df,width=0.5,color="blue")
plt.title("Sales v/s Category")
plt.show()
```

```python
#line chart
sns.lineplot(x="Category",y="Profit",data=df)
plt.title("Profit v/s Category")
plt.show()
```
