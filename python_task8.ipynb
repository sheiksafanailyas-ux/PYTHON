```python
df["Current_ratio"]=df["Current_Assets"]/df["Current_Liabilities"]
df["ROE"]=(df["Net_Income"]/df["Equity"])*100
df["EPS"]=df["Net_Income"]/df["Shares_Outstanding"]
```

```python
Max_CurrentRatio=df["Current_ratio"].max()
print("Max_Ratio:",Max_CurrentRatio)
```

```python
Max_ROE=df["ROE"].max()
print("Max_ROE:",Max_ROE)
```

```python
Max_EPS=df["EPS"].max()
print("Max_EPS:",Max_EPS)
```

```python
plt.figure(figsize=(8,5))
plt.plot(df["Year"],df["Current_ratio"],marker="o")
plt.title("Current Ratio Trend")
plt.xlabel("Year")
plt.ylabel("Current_ratio")

plt.show()
```
