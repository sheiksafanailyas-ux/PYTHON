```python
df.groupby("publisher")["recommendations"].mean().sort_values(ascending=False).head(10)
```

```python
df.groupby("is_free")["name"].count()
```

```python
df["release_year"] = pd.to_datetime(df["release_date"], errors="coerce").dt.year

year_count = df["release_year"].value_counts().sort_index()
```

```python
plt.plot(year_count.index, year_count.values)
plt.title("Games by Release Year")
plt.xlabel("Year")
plt.ylabel("Number of Games")
plt.show()
```

```python
df["publisher"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Publishers")
plt.xlabel("Publisher")
plt.ylabel("Number of Games")
plt.xticks(rotation=45)
plt.show()
```

```python
df["required_age"].value_counts().sort_index().plot(kind="bar")
plt.title("Games by Required Age")
plt.xlabel("Age")
plt.ylabel("Number of Games")
plt.show()
```

```python
plt.scatter(df["achievements"], df["recommendations"])

plt.title("Recommendations vs Achievements")
plt.xlabel("Achievements")
plt.ylabel("Recommendations")
plt.show()
```

```python
plt.hist(df["achievements"], bins=20)
plt.title("Distribution of Achievements")
plt.xlabel("Achievements")
plt.ylabel("Number of Games")
plt.show()
```

```python
plt.scatter(df["achievements"], df["recommendations"])
plt.title("Recommendations vs Achievements")
plt.xlabel("Achievements")
plt.ylabel("Recommendations")
plt.show()
```

```python
df["is_free"].value_counts().plot(kind="bar")
plt.title("Free vs Paid Games")
plt.xlabel("Free or Paid")
plt.ylabel("Number of Games")
plt.show()
```

```python
platforms = ["platforms_windows", "platforms_mac", "platforms_linux"]

df[platforms].sum().plot(kind="bar")
plt.title("Games by Platform")
plt.xlabel("Platform")
plt.ylabel("Number of Games")
plt.show()
```
