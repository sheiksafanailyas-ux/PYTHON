```python

cost_price = float(input("Enter the Cost Price: "))
selling_price = float(input("Enter the Selling Price: "))

if selling_price > cost_price:
    profit = selling_price - cost_price
    profit_percent = (profit / cost_price) * 100
    print("Profit =", profit)
    print("Profit % =", profit_percent)

elif selling_price < cost_price:
    loss = cost_price - selling_price
    loss_percent = (loss / cost_price) * 100
    print("Loss =", loss)
    print("Loss % =", loss_percent)

else:
    print("No Profit, No Loss")
```

```python
cost_price = float(input("Enter Cost Price: "))
selling_price = float(input("Enter Selling Price: "))
gst_rate = float(input("Enter GST Rate (%): "))
discount_rate = float(input("Enter Discount Rate (%): "))


if selling_price > cost_price:
    print("Profit = ₹", selling_price - cost_price)
elif selling_price < cost_price:
    print("Loss = ₹", cost_price - selling_price)
else:
    print("No Profit, No Loss")


discount = selling_price * discount_rate / 100
price_after_discount = selling_price - discount
print("Price after Discount = ", price_after_discount)

gst = price_after_discount * gst_rate / 100
final_price = price_after_discount + gst
print("Final Price after GST = ", final_price)
```
